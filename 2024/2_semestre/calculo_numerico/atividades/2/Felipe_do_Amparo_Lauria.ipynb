# **Matplotlib.pyplot e Numpy**

Para construir gráficos de uma função utilizamos **matplotlib** que permite construir diversos tipo de visualização gráfica de dados, inclusive plotar gráficos de funções elementares do Cálculo atravéz do submódulo **pyplot**. Veja https://matplotlib.org/stable/api/pyplot_summary.html  

Para escrever as funções usuais do Cálculo em python utilizamos o módulo **Numpy**. Veja https://numpy.org/doc/stable/reference/routines.math.html para as principais funções.

Outra importante função do Numpy é realizar operações matriciais utilizadas em Álgebra Linear tais como multiplicação de matrizes, cálculo de determinantes, cálculo de matrizes inversas, fatorações, etc. Veja https://numpy.org/doc/stable/reference/routines.linalg.html

Abaixo segue um exemplo de plotagem do gráfico da função exponencial y = exp(x) usando matplotlib.


```python
import numpy as np
import matplotlib.pyplot as plt


#Customização dos eixos no plano cartesiano

fig, ax = plt.subplots(figsize=(10, 10))

ax.spines['bottom'].set_position('zero')
ax.spines['left'].set_position('zero')

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

arrow_fmt = dict(markersize=4, color='black', clip_on=False)
ax.plot((1), (0), marker='>', transform=ax.get_yaxis_transform(), **arrow_fmt)
ax.plot((0), (1), marker='^', transform=ax.get_xaxis_transform(), **arrow_fmt)

#Definição da função e plotagem

x= np.linspace(-5,3,100)  #definindo o intervalo de plotagem do gŕafico
y = np.exp(x)  # definindo a função
plt.plot(x,y, color = "red", linestyle = "-", linewidth = "2")  #plotagem do gráfico

plt.show()   #exibição do gráfico
```

# **Atividade 1**

Dada uma função $f(x)$, um intervalo $[a,b]$ contendo um único zero de $f(x)$ e uma precisão $\varepsilon > 0$ , crie um código em Python que retorna uma aproximação do zero de $f(x)$ com precisão $\varepsilon$ utilizando o Método da Bisseção.

```python
# @title **Método da Bisseção**
import numpy as np

a = 1 # limite inferior do intervalo [a,b]
b = 3 # limite superior do intervalo [a,b]
p = 0.01 # precisão


#Seu código vai aqui!
def f(x):
    return x**2 - 4

#Método da bisseção
def bissecao(f, a, b, p):
  if f(a) * f(b) > 0:
    print("Intervalo inválido")
    return

  while abs(b - a) > p:
    c = (a + b) / 2
    if f(a) * f(c) < 0:
      b = c
    else:
      a = c

  return (a + b) / 2

#Solução
solucao = bissecao(f, a, b, p)

print(f"Solução aproximada: {solucao}" )
```

# **Atividade 3**

Dada uma função $f(x)$, uma aproximação inicial $x_0$ para um dos zeros de $f(x)$ e uma precisão $\varepsilon > 0$, crie um código em Python que retorna uma aproximação do zero de $f(x)$ com precisão $\varepsilon$ usando o Método de Newton-Raphson.

**Observação 1:** A precisão deve ser medida como a distãncia entre termos consecutivos da sequência de aproximações geradas pelo métodos, como fizemos durante as aulas.

**Observação 2:** O método de Newton-Raphson utiliza a derivada da função $f(x)$, cuja expressão já deve ser conhecida a priori para ser colocada no código. Outra possibilidade é utilizar o módulo **SymPy**, que serve para fazer declarar, utilizar e fazer operações com variáveis simbólicas, como por exemplo derivação e integração de funções e outras ferramentas do Cálculo. Veja https://docs.sympy.org/latest/tutorials/intro-tutorial/calculus.html  

```python
# @title **Método de Newton-Raphson**
import numpy as np
import sympy as sp  #opcional

#Definir a função f(x)
def f(x):
    return x**2 - 4

#Derivada numérica de f(x)
def df(x, h=1e-6):
    return (f(x + h) - f(x - h)) / (2 * h)

#Método de Newton-Raphson
def newton_raphson(f, df, x0, E):
    x = x0
    while abs(f(x)) > E:
        x = x - f(x) / df(x)
    return x

#Valor inicial
x0 = 1

#Precisão
E = 1e-5

solucao = newton_raphson(f, df, x0, E)

print(f"Solução aproximada: {solucao}")
```

# **Atividade 5**

Crie um código em Python que  faz a fatoração $LU$ das matrizes $A$ de tamanho $n\times n$ abaixo, sem pivoteamento parcial.

a) \begin{align}
       A = \begin{pmatrix}
        1 & 4 & 1\\
        2 & 1 & 3\\
        3 & 1 & 2
        \end{pmatrix}.
    \end{align}


b) \begin{align}
       A = \begin{pmatrix}
        -2 & 5 & 2 & 1\\
        1 & 1 & 1 & -3\\
        9 & 6 & -2 & 4\\
        -2 & 1 & 0 & 7
        \end{pmatrix}.
    \end{align}


Crie uma nova célula de código ou texto abaixo para resolver a atividade.



```python
import numpy as np

A = np.array([[1, 4, 1],
              [2, 1, 3],
              [3, 1, 2]])

#Seu código vai aqui!
def verifica_pivoteamento(A):
    if A[0, 0] != 0:
        return True
    return False

if verifica_pivoteamento(A):
    print("A matriz pode ser fatorada")

    L = np.zeros_like(A, dtype=float)
    U = np.zeros_like(A, dtype=float)
    n = A.shape[0]
    for i in range(n):
        L[i, i] = 1

        for j in range(i, n):
            U[i, j] = A[i, j] - sum(L[i, k] * U[k, j] for k in range(i))

        for j in range(i+1, n):
            L[j, i] = (A[j, i] - sum(L[j, k] * U[k, i] for k in range(i))) / U[i, i]

    print(f"L = {L}", "\n", f"U = {U}")

else:
  print("A matriz não pode ser fatorada")
```
